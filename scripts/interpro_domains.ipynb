# Data extraction from inteproscan.json

After the search in databases associated in InterProScan. We obtain a file, json file in this case, that contains the matches for the sequence, a subsequence or a substring of the requested query, in this case our K01220 consensus sequence.    


Load the file and import the necessary library in order to read de json file

```python
import json
```

```python
with open("inteproscan.json", "r", encoding="utf-8") as file:
    data = json.load(file)
```

The `key` 'matches' contains a lists with 16 dictionaries (in this particular case), each one contains the entry matches for a given sequence query.

Given a entry of the InterPro database we search it into a list of dictionaries, in this case, `data_matches`.  
We do this with the `get_entry()` function, in return we get a dictionary with the entry information and the name of the library where the entry exists.    

```python
def get_entry(entry:str, data_matches:list[dict])->dict:
    for i in data_matches:
        dict_actual = i['signature']
        accession = dict_actual['accession']
        if accession == entry:
            dict_libreria = dict_actual['signatureLibraryRelease']
            library = dict_libreria['library']
            return library, i
        #else:                       
        #    raise ValueError("entry not found")
```

For this case we are interested in the matches in the PRINTS database, so 

```python
entry = 'PR00131'
library, entry_PR00131 = get_entry(entry, data_matches)
```

We save that dictionary in `entry_PR00131`, in this case it has 6 `keys`, and the location of the regions of interest (ROIs) are in the `key` 'locations'

```python
entry_PR00131.keys()
```

```python
entry_PR00131
```

We save the ROIs locations in another variable `entry_PR00131_locs`, this new variable contains a list with lenght 4, and each elements contain information about each ROI 

```python
entry_PR00131_locs = entry_PR00131['locations']
number_of_locations = len(entry_PR00131_locs)
```

If we take a look of the locations, saved in the `entry_PR00131_locs` variable, we can see that is a list with multiple dictionaries, 4 in this case. And each dictionary represents an interval on the sequence, recognizable because the keys `start` and `end`.  

```python
entry_PR00131_locs
```

We will iterate through `entry_PR00131_locs` keys in order to save the start and end intervals inside a list. So for each interval we will have a list of a start coordinate and a end coordinate this is accomplished with `get_idx_location()` function.  
The function `get_names()` only generates a list with names for each location. 

```python
def get_idx_location(entry_PR00131_locs):
    location_start_end = [[i['start'] , i['end']] for i in entry_PR00131_locs]
    return location_start_end

def get_names(number_of_locations:int, entry):
    library_entry_num = [f'{library}_{entry}_{i}' for i in range(1, number_of_locations+1)] 
    return library_entry_num
```

```python
names_for_locations = get_names(number_of_locations, entry)
list_of_locations = get_idx_location(entry_PR00131_locs)
```
